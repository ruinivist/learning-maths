# Change of Basis

perhaps the MOST important bit of basic linear algebra, that you MUST understand intuitively

## Some notation

I should try to be consistent about my notation and setting.

Here (I) am using the standard unit basis.
$$
E = \left(\begin{bmatrix}1 \\ 0\end{bmatrix}, \begin{bmatrix}0 \\ 1\end{bmatrix}\right) = (e_0, e_1)
$$
(They) are in a different basis $B$.
$$
B = \left(\begin{bmatrix}1 \\ 1\end{bmatrix}, \begin{bmatrix}1 \\ -1\end{bmatrix}\right) = (b_0, b_1)
$$

$[v]_B$ = vector $v$ in basis $B$.

## Notion of a vector as "scalars" defining mags of an ORDERED basis

Every vector is just that, defining how it's defined in terms of a basis, such that a linear combination of that basis denotes a particular point.

In lin alg, we all agree on the origin (as the scalar would need to be $0$ in every basis) but for a "fixed" point spatially (still has to be defined in SOME basis otherwise there is no notion of "fixed"), think of the basis as the "directions". The "ordered set of scalars => vector" gives telly ou how to use that direction set. Go these units in this dir and so on. A different basis is just different direction set, using the same instruction ( instruction are define in the ordered set of direction ) will take you someplace else.

**rem: basis is direction set, a scalar is the instruction**

## A change of basis?

Encoding the same "information" in a different format. As in, how would I encode that information per an instruction/direction set defined in another basis?

Take the vector

$$
[v]_E = \begin{bmatrix}6 \\ 2\end{bmatrix}
$$

For it to be the SAME vector if I used the direction set of another guy:

$$
a\begin{bmatrix}1 \\ 1\end{bmatrix} + b\begin{bmatrix}1 \\ -1\end{bmatrix} = \begin{bmatrix}6 \\ 2\end{bmatrix},
$$
and $(a,b)$ is what I want.

Note how that is just $(\text{scalar})(\text{vec}_1) + (\text{scalar})(\text{vec}_2)$,
just a lin transform.

### What really is $B$?

The basis $B$ still needs to be defined in SOME reference; what does that $(1,1)$ mean otherwise? That $(1,1)$ is defined here in MY reference.

So the idea is, hey, this is the instruction $(6,2)$; it means $6$ in $b_1$ and then $2$ in $b_2$,
but since for you $b_1$ is $(1e_1 + 1e_2)$, you translate.

Now this $[v]_B = \begin{bmatrix}4 \\ 2\end{bmatrix}$.
Note what $B[v]_B$ gives you: $\begin{bmatrix}6 \\ 2\end{bmatrix}$.

> So when you apply that basis as a lin transform, it gives you the coords in the basis ( E ) in which
> the other basis is defined as ( here E )

$$
\underbrace{\begin{bmatrix}1 & 1 \\ 1 & -1\end{bmatrix}}_{B\text{(as in E)}}
\underbrace{\begin{bmatrix}4 \\ 2\end{bmatrix}}_{B\text{-coordinates}}
=
\underbrace{\begin{bmatrix}6 \\ 2\end{bmatrix}}_{\text{standard coordinates}}
$$

Naturally, if you wanna go the other way, you do an inverse.

**GEMS**
- when you have a different basis $B$, think about what those scalars in that $B$ are defined in; if they are in $E$, then $B$ converts to $E$ when applied. Thus this $B$ is $B(\text{theirs} \to \text{mine})$ as a transform
- $B^{-1}$ is mine to theirs
The above is also the standard convention.

## What if both bases are defined in $E$?

Note that you now have two bases $B_1$ and $B_2$, both in terms of $E$; the rules remain the same. If you have a vec $v$ in $B_1$, then $B_1v$ brings you to $E$ and then $B_2^{-1}(B_1v)$ takes you to $B_2$ ( from E ).

## What about a transformation $T$?

_here we are focused on the "matrix" representation of a transform_ => where for $B=(b_1,b_2,\ldots)$, $T = (t_1,t_2,\ldots)$ tell you where does this transform send "one unit" of $b_1$ to. Note that the numbers have to have a frame of reference, so both are tied to the same basis B, as in T makes sense when used in basis B.

Note that a transform can technically defined to be abstract like "scale long diagonal" but when you make a have to write it as a matrix, you would get different T based on what B you want to apply that transform in, to be fair, I think it makes more sense to say that transforms are define in the basis of the vector they are mean to be applied in ( if you want to go beyond english and write it as math )

Now back to the problem at hand, premise being T defines where to send one unit of basis $b_i$ to, which is $t_i$

If I want the same transformation on my vector $v_e$ (what does same mean here? same "geometrically" as in the "english sense of the transform" like double along diagonal => what effect it would have on vecs defined in $B$).

Analogy, think of basis as information encoding, $T$ as transforms in that a certain encoding. It's like software that understands a certain file format ($B$), and then can apply $T$ on that file format only. If I want to use it with a diff format, I must convert first. ( I MUST leave these analogy crutches )

$B^{-1}v_e$ would take me to their format, I apply that function that they use, $T$, and then circle back:
$$B T B^{-1} v_e$$

So the effective transformation is just $B T B^{-1}$
